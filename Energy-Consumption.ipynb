```python
import os
import pandas as pd
from dotenv import load_dotenv
from urllib.parse import quote_plus
from sqlalchemy import create_engine

load_dotenv()
password = quote_plus(os.environ["DB_PASSWORD"])
url = f"postgresql+psycopg2://{os.environ['DB_USER']}:{password}@{os.environ['DB_HOST']}:{os.environ['DB_PORT']}/{os.environ['DB_NAME']}?sslmode=require"
engine = create_engine(url)
```

```python
excelfile = pd.ExcelFile(r'E:\DEPI Professional Data Analyst\Techincal\Session 15\Energy-Consumption-SQL-Analysis\Energy Consumptions Dataset.xlsx')
excelfile.sheet_names
```

```python
energy_consumptions = pd.read_excel(excelfile, sheet_name='Energy Consumptions')
rates = pd.read_excel(excelfile, sheet_name='Rates')
buildings = pd.read_excel(excelfile, sheet_name='Building Master')
```

```python
energy_consumptions.to_sql(name = 'energy_consumptions',schema = 'energy', con = engine, if_exists = 'append', index = False)
rates.to_sql(name = 'rates',schema = 'energy', con = engine, if_exists = 'append', index = False)
buildings.to_sql(name = 'buildings',schema = 'energy', con = engine, if_exists = 'append', index = False)
```

```python
engine.dispose()
```
